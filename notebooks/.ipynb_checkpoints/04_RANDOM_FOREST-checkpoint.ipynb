# Imports & Configuration

```python
import pandas as pd
from collections import Counter
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix
)
```

```python
import os

def save_plot(fig, filename, output_dir='./outputs/random_forest'):
    os.makedirs(output_dir, exist_ok=True)
    filepath = os.path.join(output_dir, filename)
    fig.savefig(filepath, dpi=150, bbox_inches='tight', facecolor=fig.get_facecolor())
    print(f"Plot saved → {filepath}")
```

## Charger les données préparées

```python
data_path = "data/processed"

X_train = pd.read_csv(f"{data_path}/X_train.csv")
X_test  = pd.read_csv(f"{data_path}/X_test.csv")

# .squeeze() transforme DataFrame -> Series
y_train = pd.read_csv(f"{data_path}/y_train.csv").squeeze()
y_test  = pd.read_csv(f"{data_path}/y_test.csv").squeeze()

print("Données chargées avec succès")
print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
print("y_train :", y_train.shape)
print("y_test  :", y_test.shape)
```

## Calcul du class_weight

```python
counts = Counter(y_train)
total  = sum(counts.values())
class_weight = {
    0: total / (2 * counts[0]),
    1: total / (2 * counts[1])
}

print(f"Class distribution — 0: {counts[0]:,}  |  1: {counts[1]:,}")
print(f"class_weight = {class_weight}")
```

## Création du modèle Random Forest

```python
rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=5,
    min_samples_leaf=2,
    max_features='sqrt',
    class_weight=class_weight,
    random_state=42,
    n_jobs=-1
)
```

## Entraînement du modèle

```python
print("\nTraining Random Forest model...")

rf.fit(X_train, y_train)

print("Model training completed")
```

## Prédictions

```python
y_prob_rf = rf.predict_proba(X_test)[:, 1]
y_pred_rf = rf.predict(X_test)
```

## Évaluation du modèle

```python
auc_rf   = roc_auc_score(y_test, y_prob_rf)
prauc_rf = average_precision_score(y_test, y_prob_rf)

print("\n===== MODEL PERFORMANCE =====")
print(f"ROC-AUC Score : {auc_rf:.4f}")
print(f"PR-AUC Score  : {prauc_rf:.4f}")
```

## Rapport de classification

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

## Matrice de confusion

```python
import matplotlib.pyplot as plt

# Calcul de la matrice de confusion
cm = confusion_matrix(y_test, y_pred_rf)

# Création du graphique
fig, ax = plt.subplots(figsize=(6, 5))

# Affichage de la matrice
im = ax.imshow(cm)

# Ajouter les valeurs dans les cases
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(
            j, i,
            f"{cm[i, j]:,}",
            ha="center",
            va="center",
            fontsize=12
        )

# Labels axes
ax.set_xticks([0, 1])
ax.set_yticks([0, 1])

ax.set_xticklabels(["Not Fraud", "Fraud"])
ax.set_yticklabels(["Not Fraud", "Fraud"])

ax.set_xlabel("Predicted Label")
ax.set_ylabel("True Label")
ax.set_title("Confusion Matrix - Random Forest")

# Ajouter barre de couleur
plt.colorbar(im)

plt.tight_layout()
save_plot(fig, 'random_forest_confusion_matrix.png')
plt.show()
```

## Feature Importances

```python
import numpy as np

importances = rf.feature_importances_
feature_names = X_train.columns.tolist()

# Trier par importance décroissante
indices = np.argsort(importances)[::-1]
sorted_features = [feature_names[i] for i in indices]
sorted_importances = importances[indices]

# Graphique
fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.barh(sorted_features[::-1], sorted_importances[::-1],
               color='#457b9d', edgecolor='white', linewidth=0.4, alpha=0.85)

for bar, val in zip(bars, sorted_importances[::-1]):
    ax.text(val + 0.001, bar.get_y() + bar.get_height() / 2,
            f'{val:.4f}', va='center', ha='left', fontsize=8)

ax.set_xlabel("Mean Decrease in Impurity (Gini)")
ax.set_title("Feature Importances — Random Forest", fontweight='bold')
ax.grid(axis='x', alpha=0.4, linestyle='--')
plt.tight_layout()
save_plot(fig, 'random_forest_feature_importances.png')
plt.show()

print("\nTop 5 features:")
for name, imp in zip(sorted_features[:5], sorted_importances[:5]):
    print(f"  {name:<30} {imp:.4f}")
```

## Sauvegarder le modèle

```python
import os
import joblib

# Créer le dossier models s'il n'existe pas
model_dir = "models"
os.makedirs(model_dir, exist_ok=True)

# Chemin du fichier modèle
model_path = f"{model_dir}/rf_fraud_model.pkl"

# Sauvegarde du modèle
joblib.dump(rf, model_path)

print(f"\nModel saved successfully in: {model_path}")
```
