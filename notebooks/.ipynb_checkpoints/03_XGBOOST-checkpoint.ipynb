Imports & Configuration

```python
import pandas as pd
from collections import Counter
from xgboost import XGBClassifier
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix
)
```

```python
import os

def save_plot(fig, filename, output_dir='./outputs/xgboost'):
    os.makedirs(output_dir, exist_ok=True)
    filepath = os.path.join(output_dir, filename)
    fig.savefig(filepath, dpi=150, bbox_inches='tight', facecolor=fig.get_facecolor())
    print(f"Plot saved → {filepath}")
```

 Charger les données préparées

```python
data_path = "data/processed"

X_train = pd.read_csv(f"{data_path}/X_train.csv")
X_test = pd.read_csv(f"{data_path}/X_test.csv")

# .squeeze() transforme DataFrame -> Series
y_train = pd.read_csv(f"{data_path}/y_train.csv").squeeze()
y_test = pd.read_csv(f"{data_path}/y_test.csv").squeeze()

print("Données chargées avec succès")
print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
print("y_train :", y_train.shape)
print("y_test  :", y_test.shape)
```

 Calcul du scale_pos_weight

```python
scale_pos = Counter(y_train)[0] / Counter(y_train)[1]

print(f"\nScale Pos Weight = {scale_pos:.2f}")
```

Création du modèle XGBoost

```python
xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos,
    eval_metric='aucpr',
    random_state=42,
    n_jobs=-1
)
```

Entraînement du modèle

```python
print("\nTraining XGBoost model...")

xgb.fit(
    X_train,
    y_train,
    eval_set=[(X_test, y_test)],
    verbose=False
)

print("Model training completed")
```

Prédictions

```python
y_prob_xgb = xgb.predict_proba(X_test)[:, 1]
y_pred_xgb = xgb.predict(X_test)
```

 Évaluation du modèle

```python
auc_xgb = roc_auc_score(y_test, y_prob_xgb)
prauc_xgb = average_precision_score(y_test, y_prob_xgb)

print("\n===== MODEL PERFORMANCE =====")
print(f"ROC-AUC Score : {auc_xgb:.4f}")
print(f"PR-AUC Score  : {prauc_xgb:.4f}")
```

Rapport de classification

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred_xgb))
```

Matrice de confusion

```python
import matplotlib.pyplot as plt

# Calcul de la matrice de confusion
cm = confusion_matrix(y_test, y_pred_xgb)

# Création du graphique
fig, ax = plt.subplots(figsize=(6, 5))

# Affichage de la matrice
im = ax.imshow(cm)

# Ajouter les valeurs dans les cases
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        ax.text(
            j, i,
            f"{cm[i, j]:,}",
            ha="center",
            va="center",
            fontsize=12
        )

# Labels axes
ax.set_xticks([0, 1])
ax.set_yticks([0, 1])

ax.set_xticklabels(["Not Fraud", "Fraud"])
ax.set_yticklabels(["Not Fraud", "Fraud"])

ax.set_xlabel("Predicted Label")
ax.set_ylabel("True Label")
ax.set_title("Confusion Matrix - XGBoost")

# Ajouter barre de couleur
plt.colorbar(im)

plt.tight_layout()
save_plot(fig, 'xgboost_confusion_matrix.png')
plt.show()
```

Sauvegarder le modèle

```python
import os
import joblib

# Créer le dossier models s'il n'existe pas
model_dir = "models"
os.makedirs(model_dir, exist_ok=True)

# Chemin du fichier modèle
model_path = f"{model_dir}/xgb_fraud_model.pkl"

# Sauvegarde du modèle
joblib.dump(xgb, model_path)

print(f"\nModel saved successfully in: {model_path}")
```
